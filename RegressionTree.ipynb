```python
from __future__ import print_function
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt 
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import normalize
from sklearn.metrics import mean_squared_error
from sklearn.tree import DecisionTreeRegressor

 
url = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/pu9kbeSaAtRZ7RxdJKX9_A/yellow-tripdata.csv'
raw_data=pd.read_csv(url)
print(raw_data.head(9))
correlation_values=raw_data.corr()['tip_amount'].drop('tip_amount')
print(correlation_values)
abs(correlation_values).sort_values(ascending=False)[:3]
#how the input features are correlated to the target variable(tip_amount)
correlation_values.plot(kind='barh',figsize=(10,6)) 

#data preprocessing 
y=raw_data[['tip_amount']].values.astype('float32')
X=raw_data.drop(['tip_amount','store_and_fwd_flag','improvement_surcharge','VendorID','payment_type'],axis=1).values


#train/test dataset, 30% of the dataset is for testing the rest 70% for training 
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=42)

#the random_state being assigned to an integer value prevents getting a different tree every time we run the code 
dt_reg=DecisionTreeRegressor(criterion='squared_error',random_state=35,max_depth=4)
dt_reg.fit(X_train,y_train)

y_pred=dt_reg.predict(X_test)
mse_score=mean_squared_error(y_pred,y_test)
print("MSE {0:.3f}".format(mse_score))

r2_score=dt_reg.score(X_test,y_test)
print('R^2 score {0:.3f}'.format(r2_score)) 
abs(correlation_values).sort_values(ascending=False)[:3]


#changed the max_depth to 12; the results and the machine performance got worse
#removing the 4 features with 0 to little correlation to the target variable didn't change my mse and R^2 significantly as they don't have much effect on the target variable 
#changing the max_depth to 4 increased R^2 and decreased mse meaning it is a more suitable choice for this dataset
```
